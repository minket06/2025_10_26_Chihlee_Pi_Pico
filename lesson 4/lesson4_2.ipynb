```python
#條件分析
#關係運算子
#雙向選擇
# age = 17
# age = 19
age = int(input("請輸入年齡"))


if age >= 18:
    print("滿18了")
    # print("可以考駕照")
    print(f"年齡:{age}歲，可以考駕照")
else:
    print("未滿18")
    print("不可以考駕照")
    print(f"年齡:{age}歲，再等{18-age}年，就可以考駕照了")

print("應用程式結束")



```
