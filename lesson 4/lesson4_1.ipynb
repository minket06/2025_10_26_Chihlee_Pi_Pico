```python
chinese = int(input("國文分數"))
#print(type(chinese))
english = int(input("英文分數"))
#print(type(english))
math = int(input("數學分數"))
#print(type(math))

other = int(input("其他分數"))
other2 = int(input("其他分數2"))

total = chinese + english + math
total2 = chinese + english + math + other
total3 = chinese + english + math + other + other2


print("總分:",total,"分",sep="") #呼叫內建的function
print("總分2:",total2,"分",sep="") #呼叫內建的function
print("總分3:",total3,"分",sep="")

average = round(total / 3,ndigits=2)
average2 = round(total2 / 4,ndigits=2)

print("平均:",average,"分",sep="")
print("平均2:",average2,"分",sep="")

```
